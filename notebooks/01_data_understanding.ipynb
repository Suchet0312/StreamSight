# DAY 1 phase 2

```python
users_clean = users.copy()
movies_clean = movies.copy()
ratings_clean = ratings.copy()
```

```python
ratings_clean["datetime"] = pd.to_datetime(
    ratings_clean["timestamp"],
    unit="s"
)
```

```python
ratings_clean.dtypes
```

```python
ratings_clean["year"] = ratings_clean["datetime"].dt.year
ratings_clean["month"] = ratings_clean["datetime"].dt.month
ratings_clean["day"] = ratings_clean["datetime"].dt.day
ratings_clean["day_of_week"] = ratings_clean["datetime"].dt.day_name()
```

```python
ratings_clean[
    ["timestamp", "datetime", "year", "month", "day", "day_of_week"]
].head()
```

```python
movies_clean["genre"].str.split("|").head()
```

```python
movies_clean["genre_list"] = movies_clean["genre"].str.split("|")
```

```python
movies_clean[["movie_name", "genre", "genre_list"]].head()
```

```python
ratings_clean.duplicated(
    subset=["user_id", "movie_id"]
).sum()
```

```python
ratings_clean["rating"].value_counts().sort_index()
```

```python
ratings_users = ratings_clean.merge(
    users_clean,
    on="user_id",
    how="left"
)
```

```python
ratings_users.head()
```

```python
stream_data = ratings_users.merge(
    movies_clean,
    on="movie_id",
    how="left"
)
```

```python
stream_data.head()
```

```python
stream_data.shape
```

```python
stream_data.columns
```

```python
stream_data[
    [
        "user_id",
        "gender",
        "age",
        "occupation",
        "movie_id",
        "movie_name",
        "genre",
        "rating",
        "datetime"
    ]
].head(10)
```

```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("../streamsight.db")
```

```python
users_clean.to_sql(
    "users",
    conn,
    if_exists="replace",
    index=False
)

movies_clean.drop(columns=["genre_list"]).to_sql(
    "movies",
    conn,
    if_exists="replace",
    index=False
)

ratings_clean.to_sql(
    "ratings",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)
```

```python
pd.read_sql(
    """
    SELECT COUNT(*) AS total_users
    FROM users;
    """,
    conn
)
```
